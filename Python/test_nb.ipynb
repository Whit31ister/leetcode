```python
target_ip = "192.168.1.1"      # String
open_ports = [22, 80, 443]     # List
is_vulnerable = True           # Boolean

target_info = {
    "ip": target_ip,
    "os": "Linux",
    "ports": open_ports
}

print(f"Scanning target: {target_info['ip']}")
```

```python
import socket

def check_port(ip, port):
    try:
        # Create a socket object (AF_INET = IPv4, SOCK_STREAM = TCP)
        s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        s.settimeout(1.0) # Set timeout to 1 second
        
        result = s.connect_ex((ip, port)) # Returns 0 if connection succeeds
        s.close()
        
        return result == 0
    except Exception as e:
        print(f"Error checking port {port}: {e}")
        return False

# Usage
status = check_port("127.0.0.1", 80)
print(f"Port 80 status: {'Open' if status else 'Closed'}")
```
